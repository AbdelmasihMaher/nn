```python
import numpy as np
import matplotlib.pyplot as plt

from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten, Dropout, BatchNormalization
from tensorflow.keras.utils import to_categorical
```

```python
(train_images, train_labels), (test_images, test_labels) = mnist.load_data()

print("Training Images Shape:", train_images.shape)
print("Testing Images Shape:", test_images.shape)

# Show Sample Image
plt.imshow(train_images[0], cmap='gray')
plt.title("Real Label: " + str(train_labels[0]))
plt.show()
```

```python
# 3. Data Preprocessing

# Normalize images
train_images = train_images / 255.0
test_images = test_images / 255.0

# Convert labels to categorical
train_labels = to_categorical(train_labels, 10)
test_labels = to_categorical(test_labels, 10)
```

```python
# 4. Build MLP Model Function
def build_mlp_model(activation_function='relu'):
    model = Sequential()

    # Convert image to vector
    model.add(Flatten(input_shape=(28, 28)))

    # First Hidden Layer
    model.add(Dense(128, activation=activation_function))
    model.add(BatchNormalization())
    model.add(Dropout(0.3))

    # Second Hidden Layer
    model.add(Dense(64, activation=activation_function))

    # Output Layer
    model.add(Dense(10, activation='softmax'))

    return model
```

```python
# 5. Experimentation Requirements (Mandatory)

#Experiment 1: ReLU Activation (Model A)
print("\n=== Starting Experiment 1: ReLU Activation ===")
model_relu = build_mlp_model(activation_function='relu')
model_relu.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

training_history_relu = model_relu.fit(
    train_images,
    train_labels,
    epochs=10,
    batch_size=32,
    validation_split=0.2
)

test_loss_relu, test_accuracy_relu = model_relu.evaluate(test_images, test_labels, verbose=0)


#  Experiment 2: Sigmoid Activation (Model B)
print("\n=== Starting Experiment 2: Sigmoid Activation ===")
model_sigmoid = build_mlp_model(activation_function='sigmoid')
model_sigmoid.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

training_history_sigmoid = model_sigmoid.fit(
    train_images,
    train_labels,
    epochs=10,
    batch_size=32,
    validation_split=0.2
)

test_loss_sigmoid, test_accuracy_sigmoid = model_sigmoid.evaluate(test_images, test_labels, verbose=0)
```

```python
# 6. Visualization Requirements
def plot_curves(history, title):
    # Accuracy Graph
    plt.figure(figsize=(12, 4))
    plt.subplot(1, 2, 1)
    plt.plot(history.history['accuracy'])
    plt.plot(history.history['val_accuracy'])
    plt.title(f'{title} - Training vs Validation Accuracy')
    plt.xlabel('Epoch')
    plt.ylabel('Accuracy')
    plt.legend(['Train', 'Validation'])

    # Loss Graph
    plt.subplot(1, 2, 2)
    plt.plot(history.history['loss'])
    plt.plot(history.history['val_loss'])
    plt.title(f'{title} - Training vs Validation Loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend(['Train', 'Validation'])
    plt.show()

# Show graphs for both experiments
plot_curves(training_history_relu, "Model A (ReLU)")
plot_curves(training_history_sigmoid, "Model B (Sigmoid)")
```

```python
# 7. Prediction Example (Using Best Model)

predictions = model_relu.predict(test_images)

plt.imshow(test_images[0], cmap='gray')
plt.title("Predicted Label: " + str(np.argmax(predictions[0])))
plt.show()
```

```python
# 8. Results Requirements (Comparison Table)
print("\n==================================================")
print(f"{'Model':<20} | {'Accuracy':<15} | {'Loss':<10}")
print("--------------------------------------------------")
print(f"{'Model A (ReLU)':<20} | {test_accuracy_relu*100:.2f}%{'':<9} | {test_loss_relu:.4f}")
print(f"{'Model B (Sigmoid)':<20} | {test_accuracy_sigmoid*100:.2f}%{'':<9} | {test_loss_sigmoid:.4f}")
print("==================================================")
```
